```python
plt.figure(figsize=(8, 6))
species = iris_df['Iris_type'].unique()
colors = ['red', 'green', 'blue']
# Scatter plot for each species
for i in range(len(species)):
    x = iris_df[iris_df['Iris_type'] == species[i]]
    plt.scatter(x['sepal length (cm)'], x['sepal width (cm)'], c=colors[i], label=species[i])
# Add labels and legend
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.title('Sepal Length vs Width by Species')
plt.legend()
# Show plot
plt.show()

iris_df.plot(kind="scatter", x="sepal length (cm)", y="sepal width (cm)")
```
